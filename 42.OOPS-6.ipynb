```python
class Father:

    def __init__(self, fname, lname):
        self.fname = fname
        self.lname = lname
        
class Child(Father):

    def __init__(self, fname, lname, cname, cage):
        super().__init__(fname, lname)
        self.cname = cname
        self.cage = cage

    def show_attrs(self):
        for key, value in self.__dict__.items():
            print(f"{key} : {value}")
```

```python
c1 = Child('ram', 'sharma', 'aman', 20)
```

```python
class Father:

    def __init__(self, fname, lname):
        self.fname = fname
        self.lname = lname

class Mother:
    
    def __init__(self, mname, lname):
        self.mname = mname
        self.lname = lname
        
class Child(Father,Mother):         # MRO   --->  Left to right

    def __init__(self, fname, lname, mname, cname, cage):
        super().__init__(fname, lname)
        Mother.__init__(self,mname, lname)
        self.cname = cname
        self.cage = cage

    def show_attrs(self):
        print(f"Child name: {self.cname} {self.lname} Child age: {self.cage}")
        print(f"Father's name: {self.fname} {self.lname}")
        print(f"Mother's name: {self.mname} {self.lname}")
```

```python
c1 = Child('ram', 'sharma', 'simran', 'aman', 23)
```

```python
c1.show_attrs()
```

## Hierarchical Inheritance


            parent
          /       \
        child1   child2

```python
class Animal:

    def __init__(self, type_, color):
        self.type_ = type_
        self.color = color

    def show_attrs(self):
        for key, value in self.__dict__.items():
            print(f"{key} : {value}")

    def change_attrs(self):
        for i in self.__dict__:
            self.__dict__[i] = input(f'Enter new value for {i}: ')
            
class Cat(Animal):
    
    def __init__(self, name, type_, color):
        super().__init__(type_, color)
        self.name = name

class Dog(Animal):

    def __init__(self, name, type_, color):
        super().__init__(type_, color)
        self.name = name
```

```python
c1 = Cat('kitten', 'street', 'brown')
d1 = Dog('sheru', 'street', 'black')
```

```python
c1.show_attrs()
```

```python
d1.show_attrs()
```

```python
c1.change_attrs()
```

```python
c1.show_attrs()
```

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __lt__(self, value):
        if self.age < value.age:
            return True
        else:
            return False
```

```python
p1 = Person('rohan', 23)
p2 = Person('aman', 22)
```

```python
p1 < p2
```
