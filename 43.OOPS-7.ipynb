# Access Specifiers

                    1. Public          --> varname

                    2. Protected       -->  _varname

                    3. Private         -->  __varname

```python
class Specifiers:

    public = "public variable"
    _pro = "protected variable"
    __pri = "private variable"

    def show_attrs(self):
        print(f"Public: {Specifiers.public}")
        print(f"Protected: {Specifiers._pro}")
        print(f"Private: {Specifiers.__pri}")

    def change_attrs(self):
        Specifiers.public = input('Enter new public: ')
        Specifiers._pro = input('Enter new protected: ')
        Specifiers.__pri = input('Enter new private: ')
        
```

```python
s1 = Specifiers()
```

```python
s1.show_attrs()
```

```python
s1.change_attrs()
```

```python
s1.show_attrs()
```

```python
s1.public
```

```python
s1._pro
```

```python
s1.__pri
```

```python
class Name:

    __slots__ = ('__firstname', '__lastname')
    def __init__(self, fname, lname):
        self.__firstname = fname
        self.__lastname = lname

class Student(Name):

    __slots__ = ['__fees']
    def __init__(self, fname, lname,fees):
        super().__init__(fname,lname)
        self.__fees = fees
```
