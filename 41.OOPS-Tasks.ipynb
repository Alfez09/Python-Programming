        Create a Student class and initialize it with name and roll number. Make methods to :         
1. Display - It should display all informations of the student.         
2. setAge - It should assign age to student         
3. setMarks - It should assign marks to the studen 62 minute. 

```python
class Student:

    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
    def Display(self):
        for key, value in self.__dict__.items():
            print(f"{key} : {value}")
            
    def setAge(self, age):
        self.age = age
        
    def setMarks(self, *marks):
        self.marks = marks
```

```python
s1 = Student('rahul', 20)
s2 = Student('nitin', 25)
```

```python
class Person:

    def __init__(self, lname, fname, phoneno, email):
        self.lname = lname
        self.fname = fname
        self.phoneno = phoneno
        self.email = email

    def __str__(self):
        data = f"{self.lname}, {self.fname} -- Phone Number: {self.phoneno} -- Email Address(es): "
        if type(self.email) == list:
            data += ', '.join(self.email)
            return data
        else:
            return data + self.email
```

```python
class AddressBook(Person):

    def __init__(self):
        self.data = dict()

    def add_person(self, obj):
        if len(self.data) == 0:
            self.data[1] = obj
        else:
            v = list(self.data.keys())[-1] + 1
            self.data[v] = obj
    
    def lookup_contact(self, lname, fname=None):
        if fname:
            for obj in self.data.values():
                if obj.lname == lname and obj.fname == fname:
                    print(obj)
        else:
            for obj in self.data.values():
                if obj.lname == lname:
                    print(obj)
                
```

```python
a = AddressBook()
```

```python
a.add_person(Emily)
a.add_person(Wesley)
```

```python
a.lookup_contact('Zhang')
```
