```python
import requests


def get_place_data(query, lat, lon, sort="RELEVANCE", limit=10):
    api_key = "fsq3KyMn2pQyw4/YG3Gpt9halKQ5P362Tv3l2c1MwvS5WZg="
    
    url = "https://api.foursquare.com/v3/places/search"

    ll = f"{lat},{lon}"
    
    dict1 = {
        'query':query,
        'll': ll,
        'sort':sort,
        'limit':limit
    }
    
    headers = {
        "accept": "application/json",
        "Authorization": api_key
    }
    
    resp = requests.get(url, params=dict1, headers=headers)
    if resp.status_code == 200:
        data = resp.json()
        for i in data['results']:
            name = i['name']
            address = i['location']['formatted_address']
            print(f"Place_name: {name}")
            print(f"Place_address: {address}")
            print()
            print()
    else:
        print(f"Error occured status code: {resp.status_code} Reason: {resp.reason}")
```
