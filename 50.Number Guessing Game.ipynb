```python
import random
```

```python
computer_guess = random.randint(1, 50)

chance = 1

while chance <= 5:
    user_guess = int(input(f'Chance[{chance}] ---> Enter user guess(1-50): '))
    
    if user_guess < 1 or user_guess > 50:
        print('Guess is out of range.')
        print('Please enter a number btw 1 - 50')
        continue
    
    if user_guess == computer_guess:
        print('Winner...')
        break
    elif user_guess > computer_guess:
        print('User guess is high. Please enter a small number')
    elif user_guess < computer_guess:
        print('User guess is small. Please enter a big number')
        
        
    chance += 1
else:
    print('Looser...')
    print(f'Computer guess was: {computer_guess}')

  
```
