    4. Variable Length Argument

```python
def fun(a,b):
    return a/ b
```

```python
fun(5,6)
```

    fun --> argument --> number
    
            return True --> number even
            
            return False --> number odd

```python
def even_odd(num):
    """return True if the num is even otherwise return False"""
    if num % 2 == 0:
        return True
    else:
        return False
```

```python
even_odd(5)
```

```python
even_odd(10)
```

```python
print(even_odd.__doc__)
```

## Required / Positional Argument

```python
def user_details(name, age, address):   # formal arguments
    print(f"User_name: {name}")
    print(f"User_age: {age}")
    print(f"User_address: {address}")
```

```python
user_details('harsh', 25, 'jaipur')  # actual argument
```

```python
user_details('aman', 24)
```

```python
user_details()
```

```python
user_details(25, 'jaipur', 'rahul')
```

## keyword argument

```python
def user_details(name, age, address):   # formal arguments
    print(f"User_name: {name}")
    print(f"User_age: {age}")
    print(f"User_address: {address}")
```

```python
user_details(25, 'aman', 'jaipur')
```

```python
user_details(age=25, name='rahul', address='jaipur')
```

```python
user_details(name='rohan', 25, address='jaipur')
```

```python
user_details('rohan', age=25, address='jaipur')
```

```python
user_details(name='ashish', 'vaibhav', 25, 'ajmer')
```

```python
user_details('ashish', name='vaibhav', age=25, address='ajmer')
```
